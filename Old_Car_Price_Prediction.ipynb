```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder,OneHotEncoder,OrdinalEncoder
from sklearn.preprocessing import Normalizer,StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
```

```python
df=pd.read_csv(r"C:\Users\Amit Kumar\Desktop\ML\LR_Model\used_car_price_prediction.csv")
df
df.shape
```

```python
df.describe()
df.info()
df.head()
df.tail()
df.isnull()
df.isnull().sum()
df["Transmission"].unique()
```

```python
label=LabelEncoder()
df["Encoded_Transmission"]=label.fit_transform(df["Transmission"])
df
```

```python
onehot=OneHotEncoder()
v1=onehot.fit_transform(df[["Fuel_Type"]])
Encoded_value=v1.toarray()
col_name=onehot.get_feature_names_out()
encoded_df=pd.DataFrame(Encoded_value,columns=col_name,index=df.index)
encoded_df
df=pd.concat([df,encoded_df],axis=1)
df
df=df.drop(columns=["Fuel_Type"])
df=df.drop(columns=["Transmission"])
df
```

```python
# Step 1: Sirf un 4 numerical columns ki list banayein
num_cols = [
    "Car_Age_Years",
    "Kilometres_Driven",
    "Engine_Capacity_CC",
    "Selling_Price_INR",
]

# Step 2: Sirf unhi 4 columns par Normalizer lagayein
NS = Normalizer()
v = NS.fit_transform(df[num_cols])  # <-- yahan sirf 4 columns pass kiye

# Step 3: DataFrame banayein
df2 = pd.DataFrame(v, columns=num_cols)
```

```python
# Purane unscaled columns ko df2 (scaled values) se replace kar do
df[num_cols] = df2
df
```

```python
x=df.drop("Selling_Price_INR",axis=1)
x
y=df["Selling_Price_INR"]
y
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
LR=LinearRegression()
LR.fit(x_train,y_train)
```

```python
pred=LR.predict(x_test)
pred
```

```python
r2=r2_score(y_test,pred)
mse=mean_squared_error(y_test,pred)
mae=mean_absolute_error(y_test,pred)
print("r2_score : ",r2)
print("mean_squared_error : ",mse)
print("mean_absolute_error : ",mae)
```
